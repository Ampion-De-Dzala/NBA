#### **Importation des données**

<small style="line-height: 1.5">
Les données sont disponibles sur le site de Kaggle. Les champs sont les suivants :

- **Rk** : rang ou identifiant du joueur dans les statistiques de la saison  
- **Player** : nom du joueur  
- **Pos** : poste principal du joueur (PG : meneur, SG : arrière, SF : ailier, PF : ailier fort, C : pivot)  
- **Age** : âge du joueur durant la saison  
- **Tm** : code de l'équipe NBA du joueur  
- **G** : nombre de matchs joués  
- **GS** : nombre de matchs débutés en tant que titulaire  
- **MP** : moyenne de minutes jouées par match  
- **FG** : moyenne de tirs réussis par match (Field Goals)  
- **FGA** : moyenne de tirs tentés par match  
- **FG%** : pourcentage de réussite aux tirs  
- **3P** : moyenne de tirs à trois points réussis par match  
- **3PA** : moyenne de tirs à trois points tentés par match  
- **3P%** : pourcentage de réussite à trois points  
- **2P** : moyenne de tirs à deux points réussis par match  
- **2PA** : moyenne de tirs à deux points tentés par match  
- **2P%** : pourcentage de réussite à deux points  
- **eFG%** : pourcentage effectif de réussite aux tirs (Effective Field Goal Percentage)  
- **FT** : moyenne de lancers francs réussis par match  
- **FTA** : moyenne de lancers francs tentés par match  
- **FT%** : pourcentage de réussite aux lancers francs  
- **ORB** : moyenne de rebonds offensifs par match  
- **DRB** : moyenne de rebonds défensifs par match  
- **TRB** : moyenne totale de rebonds par match  
- **AST** : moyenne de passes décisives par match  
- **STL** : moyenne d'interceptions par match  
- **BLK** : moyenne de contres par match  
- **TOV** : moyenne de pertes de balle par match  
- **PF** : moyenne de fautes personnelles par match  
- **PTS** : moyenne de points marqués par match  
- **Season** : saison NBA concernée (exemple : 1997-98)  
- **MVP** : indicateur du joueur élu MVP de la saison (True = MVP, False = non MVP)

Il contient 14 573 lignes représentant les statistiques annuelles de joueurs NBA par saison, incluant les moyennes par match pour les tirs, rebonds, passes décisives, interceptions, contres, et un indicateur si le joueur a été élu MVP.

<small style="line-height: 1.5">
Les valeurs manquantes observées dans les colonnes de pourcentage (3P%, FT%, 2P%, FG%, eFG%) ne correspondent pas à des données absentes, mais à des cas où le joueur n’a effectué aucune tentative dans la catégorie concernée.
Dans ces situations, le pourcentage est mathématiquement indéfini (division par zéro).
Ces valeurs sont donc conservées en NaN afin de respecter la structure réelle des données et d’éviter toute interprétation incorrecte.</small>

```python
from openpyxl import load_workbook
from openpyxl.styles import Font, PatternFill, Alignment, Border, Side, GradientFill
from openpyxl.utils import get_column_letter
from openpyxl.worksheet.datavalidation import DataValidation
from openpyxl.chart import BarChart, LineChart, RadarChart, Reference, PieChart, BubbleChart, ScatterChart
from openpyxl.chart.label import DataLabelList
from openpyxl.chart.series import SeriesLabel
from openpyxl.chart.data_source import NumDataSource, NumRef
from openpyxl.drawing.fill import PatternFillProperties

# ═══════════════════════════════════════════════════════════════════════════════
# PALETTE & STYLES
# ═══════════════════════════════════════════════════════════════════════════════
wb = load_workbook(output_path)

# Palette NBA premium
C_MIDNIGHT  = "0D1B2A"
C_NAVY      = "1F4E78"
C_STEEL     = "2E75B6"
C_GOLD      = "C9A84C"
C_GOLD_LIGHT= "F4D03F"
C_GREEN     = "1E8449"
C_TEAL      = "148F77"
C_RED       = "C0392B"
C_ORANGE    = "E67E22"
C_LIGHT_BG  = "EBF5FB"
C_LIGHT_HDR = "D6EAF8"
C_SEPARATOR = "AED6F1"
C_WHITE     = "FFFFFF"

def fill(hex_color):
    return PatternFill(start_color=hex_color, end_color=hex_color, fill_type="solid")

def font(bold=False, size=10, color="000000", name="Arial", italic=False):
    return Font(bold=bold, size=size, color=color, name=name, italic=italic)

def border_full(style='thin'):
    s = Side(style=style)
    return Border(left=s, right=s, top=s, bottom=s)

def border_bottom(style='medium'):
    return Border(bottom=Side(style=style))

def align(h='center', v='center', wrap=False):
    return Alignment(horizontal=h, vertical=v, wrap_text=wrap)

thin = border_full('thin')
medium = border_full('medium')

# Styles nommés
S_TITLE     = {'fill': fill(C_MIDNIGHT), 'font': font(True, 18, C_WHITE, 'Arial')}
S_SECTION   = {'fill': fill(C_NAVY),     'font': font(True, 11, C_WHITE, 'Arial')}
S_SUBHDR    = {'fill': fill(C_LIGHT_HDR), 'font': font(True,  9, C_NAVY,  'Arial')}
S_KPI_HDR   = {'fill': fill(C_STEEL),    'font': font(True, 10, C_WHITE, 'Arial')}
S_KPI_VAL   = {'fill': fill(C_LIGHT_BG), 'font': font(True, 15, C_NAVY,  'Arial')}
S_KPI_NAME  = {'fill': fill(C_WHITE),    'font': font(False, 9, C_MIDNIGHT, 'Arial')}
S_KPI_SEAS  = {'fill': fill(C_WHITE),    'font': font(False, 8, "7F8C8D", 'Arial', italic=True)}
S_FILTER_LB = {'fill': fill("E8F4F8"),   'font': font(True,  9, C_NAVY,  'Arial')}
S_DATA_ODD  = {'fill': fill(C_WHITE),    'font': font(False, 9, C_MIDNIGHT)}
S_DATA_EVEN = {'fill': fill("F2F9FF"),   'font': font(False, 9, C_MIDNIGHT)}

# Mise à jour du dictionnaire COL avec les nouvelles lettres de colonnes
COL = {
    "Player": "B", "Season": "AE", "Tm": "E", "Pos": "C", "G": "F",
    "PTS": "AD",    "AST": "Y",   "TRB": "X", "STL": "Z", "BLK": "AA",
    "eFG%": "R",   "3P%": "N",  "3PA": "M",  "MVP": "AF",
}

DR = "'Données brutes'"
P, S_col = COL["Player"], COL["Season"]
AST, TRB, STL, BLK = COL["AST"], COL["TRB"], COL["STL"], COL["BLK"]
EFG, _3P, _3PA, MVP_c = COL["eFG%"], COL["3P%"], COL["3PA"], COL["MVP"]

def apply_style(cell, s):
    for attr, val in s.items():
        setattr(cell, attr, val)

def write_section_header(ws, cell, merge_range, title, style=None):
    """Merge EN PREMIER, style uniquement sur cellule principale."""
    ws.merge_cells(merge_range)
    c = ws[cell]
    c.value = title
    s = style or S_SECTION
    apply_style(c, s)
    c.alignment = align('center', 'center')
    c.border = medium

# ═══════════════════════════════════════════════════════════════════════════════
# FEUILLE TDB_NBA
# ═══════════════════════════════════════════════════════════════════════════════
if "TDB_NBA" in wb.sheetnames:
    del wb["TDB_NBA"]
ws = wb.create_sheet("TDB_NBA", 0)
ws.page_setup.fitToPage = True
ws.sheet_view.showGridLines = False

# ── LIGNE 1 : Bandeau titre principal ─────────────────────────────────────────
ws.merge_cells('A1:N2')
c = ws['A1']
c.value = "🏀  DASHBOARD NBA — STATISTIQUES ALL-TIME"
apply_style(c, S_TITLE)
c.alignment = align('center', 'center')
c.border = medium
ws.row_dimensions[1].height = 22
ws.row_dimensions[2].height = 22

# Ligne décorative dorée sous le titre
ws.merge_cells('A3:N3')
ws['A3'].fill = fill(C_GOLD)
ws.row_dimensions[3].height = 4

# ── LIGNE 4-5 : Sous-titre et date ────────────────────────────────────────────
ws.merge_cells('A4:J4')
ws['A4'].value = "Vue d'ensemble des meilleures performances de l'histoire NBA par joueur et par saison"
ws['A4'].font = font(False, 9, "5D6D7E", 'Arial', italic=True)
ws['A4'].alignment = align('left', 'center')
ws.row_dimensions[4].height = 16

# ── LIGNE 6 : SECTION FILTRES ─────────────────────────────────────────────────
ws.row_dimensions[5].height = 8

write_section_header(ws, 'A6', 'A6:N6', "▼  FILTRES INTERACTIFS — Utilisez les listes déroulantes pour explorer les données")
ws.row_dimensions[6].height = 18

# Labels + dropdowns sur ligne 7
def make_list_dv(values):
    joined = ",".join(str(v) for v in sorted(values))
    if len(joined) > 250:
        joined = joined[:250].rsplit(",", 1)[0]
    return DataValidation(type="list", formula1=f'"{joined}"', allow_blank=True)

dv_season = make_list_dv(df['Season'].dropna().unique())
dv_pos = make_list_dv(df['Pos'].dropna().unique())
dv_team = make_list_dv(df['Tm'].dropna().unique())
dv_mvp = DataValidation(type="list", formula1='"Oui,Non"', allow_blank=True)

filter_items = [
    ("A7", "B7", "🗓  Saison :", dv_season),
    ("D7", "E7", "📍 Poste :", dv_pos),
    ("G7", "H7", "🏟  Équipe :", dv_team),
    ("J7", "K7", "🏆 MVP :", dv_mvp),
]
ws.row_dimensions[7].height = 20
for lbl_cell, val_cell, label, dv in filter_items:
    c = ws[lbl_cell]
    c.value = label
    apply_style(c, S_FILTER_LB)
    c.alignment = align('right', 'center')
    c.border = thin
    ws[val_cell].fill = fill(C_WHITE)
    ws[val_cell].border = Border(bottom=Side(style='medium', color=C_GOLD))
    ws[val_cell].alignment = align('center', 'center')
    ws[val_cell].font = font(True, 9, C_NAVY)
    dv.add(ws[val_cell])
    ws.add_data_validation(dv)

# ── ESPACE ────────────────────────────────────────────────────────────────────
ws.row_dimensions[8].height = 10

# ── LIGNE 9 : SECTION KPI ─────────────────────────────────────────────────────
write_section_header(ws, 'A9', 'A9:N9', "★  RECORDS ALL-TIME — Les meilleures performances de l'histoire")
ws.row_dimensions[9].height = 18

# KPI definitions (titre, formule_val, formule_player, formule_saison)
kpi_defs = [
    (
        "MEILLEUR\nSCOREUR", "PTS",
        f"=MAX({DR}!{COL['PTS']}:{COL['PTS']})",
        f"=INDEX({DR}!{P}:{P},MATCH(MAX({DR}!{COL['PTS']}:{COL['PTS']}),{DR}!{COL['PTS']}:{COL['PTS']},0))",
        f"=INDEX({DR}!{S_col}:{S_col},MATCH(MAX({DR}!{COL['PTS']}:{COL['PTS']}),{DR}!{COL['PTS']}:{COL['PTS']},0))",
    ),
    (
        "MEILLEUR\nPASSEUR", "AST",
        f"=MAX({DR}!{AST}:{AST})",
        f"=INDEX({DR}!{P}:{P},MATCH(MAX({DR}!{AST}:{AST}),{DR}!{AST}:{AST},0))",
        f"=INDEX({DR}!{S_col}:{S_col},MATCH(MAX({DR}!{AST}:{AST}),{DR}!{AST}:{AST},0))",
    ),
    (
        "MEILLEUR\nREBONDEUR", "TRB",
        f"=MAX({DR}!{TRB}:{TRB})",
        f"=INDEX({DR}!{P}:{P},MATCH(MAX({DR}!{TRB}:{TRB}),{DR}!{TRB}:{TRB},0))",
        f"=INDEX({DR}!{S_col}:{S_col},MATCH(MAX({DR}!{TRB}:{TRB}),{DR}!{TRB}:{TRB},0))",
    ),
    (
        "MEILLEUR\nINTERCEPTEUR", "STL",
        f"=MAX({DR}!{STL}:{STL})",
        f"=INDEX({DR}!{P}:{P},MATCH(MAX({DR}!{STL}:{STL}),{DR}!{STL}:{STL},0))",
        f"=INDEX({DR}!{S_col}:{S_col},MATCH(MAX({DR}!{STL}:{STL}),{DR}!{STL}:{STL},0))",
    ),
    (
        "MEILLEUR\nCONTREUR", "BLK",
        f"=MAX({DR}!{BLK}:{BLK})",
        f"=INDEX({DR}!{P}:{P},MATCH(MAX({DR}!{BLK}:{BLK}),{DR}!{BLK}:{BLK},0))",
        f"=INDEX({DR}!{S_col}:{S_col},MATCH(MAX({DR}!{BLK}:{BLK}),{DR}!{BLK}:{BLK},0))",
    ),
    (
        "MEILLEURE\nEFFICACITÉ", "eFG%",
        f"=MAX({DR}!{EFG}:{EFG})",
        f"=INDEX({DR}!{P}:{P},MATCH(MAX({DR}!{EFG}:{EFG}),{DR}!{EFG}:{EFG},0))",
        f"=INDEX({DR}!{S_col}:{S_col},MATCH(MAX({DR}!{EFG}:{EFG}),{DR}!{EFG}:{EFG},0))",
    ),
    # Meilleur shooteur 3P (max de % sans condition)
    (
    "MEILLEUR\nSHOOTEUR 3P", "3P",  # <-- Changé en "3P" au lieu de "3P%"
    f"=MAX({DR}!{_3PA}:{_3PA})",  # <-- Utilise _3PA (nombre de tirs) au lieu de _3P (pourcentage)
    f"=INDEX({DR}!{P}:{P},MATCH(MAX({DR}!{_3PA}:{_3PA}),{DR}!{_3PA}:{_3PA},0))",
    f"=INDEX({DR}!{S_col}:{S_col},MATCH(MAX({DR}!{_3PA}:{_3PA}),{DR}!{_3PA}:{_3PA},0))",
    ),
    # Joueur le plus complet
    (
        "JOUEUR LE\nPLUS COMPLET", "Score",
        f"=MAX({DR}!{COL['PTS']}:{COL['PTS']}+1.5*{DR}!{AST}:{AST}+1.25*{DR}!{TRB}:{TRB}+2*{DR}!{STL}:{STL}+2*{DR}!{BLK}:{BLK})",
        f"=INDEX({DR}!{P}:{P},MATCH(MAX({DR}!{COL['PTS']}:{COL['PTS']}+1.5*{DR}!{AST}:{AST}+1.25*{DR}!{TRB}:{TRB}+2*{DR}!{STL}:{STL}+2*{DR}!{BLK}:{BLK}),{DR}!{COL['PTS']}:{COL['PTS']}+1.5*{DR}!{AST}:{AST}+1.25*{DR}!{TRB}:{TRB}+2*{DR}!{STL}:{STL}+2*{DR}!{BLK}:{BLK},0))",
        f"=INDEX({DR}!{S_col}:{S_col},MATCH(MAX({DR}!{COL['PTS']}:{COL['PTS']}+1.5*{DR}!{AST}:{AST}+1.25*{DR}!{TRB}:{TRB}+2*{DR}!{STL}:{STL}+2*{DR}!{BLK}:{BLK}),{DR}!{COL['PTS']}:{COL['PTS']}+1.5*{DR}!{AST}:{AST}+1.25*{DR}!{TRB}:{TRB}+2*{DR}!{STL}:{STL}+2*{DR}!{BLK}:{BLK},0))",
    ),
    # Plus de MVPs (compte des "True" dans la colonne MVP)
    (
    "PLUS DE\nMVPs", "Nb MVP",
    f"=MAX(COUNTIFS({DR}!{P}:{P},{DR}!{P}:{P},{DR}!{MVP_c}:{MVP_c},\"True\"))",
    f"=INDEX({DR}!{P}:{P},MATCH(MAX(COUNTIFS({DR}!{P}:{P},{DR}!{P}:{P},{DR}!{MVP_c}:{MVP_c},\"True\")),COUNTIFS({DR}!{P}:{P},{DR}!{P}:{P},{DR}!{MVP_c}:{MVP_c},\"True\"),0))",
    f"=IFERROR(TEXTJOIN(\", \",True,IF({DR}!{P}:{P}=INDEX({DR}!{P}:{P},MATCH(MAX(COUNTIFS({DR}!{P}:{P},{DR}!{P}:{P},{DR}!{MVP_c}:{MVP_c},\"True\")),COUNTIFS({DR}!{P}:{P},{DR}!{P}:{P},{DR}!{MVP_c}:{MVP_c},\"True\"),0)),{DR}!{S_col}:{S_col},\"\")),\"\")",
    ),
]

# Couleurs alternées pour les KPI cards
kpi_colors = [C_NAVY, C_STEEL, C_TEAL, C_GREEN, C_RED, C_ORANGE, C_NAVY, C_STEEL, C_TEAL]

ws.row_dimensions[10].height = 32
ws.row_dimensions[11].height = 28
ws.row_dimensions[12].height = 22
ws.row_dimensions[13].height = 16
ws.row_dimensions[14].height = 16

for idx, (titre, metric_lbl, f_val, f_player, f_saison) in enumerate(kpi_defs):
    col_letter = get_column_letter(idx + 1)
    kpi_color = kpi_colors[idx]

    # Ligne 10 : titre KPI
    c10 = ws[f'{col_letter}10']
    c10.value = titre
    c10.fill = fill(kpi_color)
    c10.font = font(True, 9, C_WHITE, 'Arial')
    c10.alignment = align('center', 'center', wrap=True)
    c10.border = medium

    # Ligne 11 : valeur
    c11 = ws[f'{col_letter}11']
    c11.value = f_val
    c11.fill = fill(C_LIGHT_BG)
    c11.font = Font(bold=True, size=14, color=kpi_color, name='Arial')
    c11.alignment = align('center', 'center')
    c11.border = thin

    # Ligne 12 : nom du joueur
    c12 = ws[f'{col_letter}12']
    c12.value = f_player
    c12.fill = fill(C_WHITE)
    c12.font = font(True, 8, C_MIDNIGHT, 'Arial')
    c12.alignment = align('center', 'center', wrap=True)
    c12.border = thin

    # Ligne 13 : saison
    c13 = ws[f'{col_letter}13']
    c13.value = f_saison
    c13.fill = fill(C_WHITE)
    c13.font = font(False, 8, "95A5A6", 'Arial', italic=True)
    c13.alignment = align('center', 'center')
    c13.border = thin

    # Ligne 14 : badge métrique
    c14 = ws[f'{col_letter}14']
    c14.value = metric_lbl
    c14.fill = fill(kpi_color)
    c14.font = font(True, 7, C_WHITE, 'Arial')
    c14.alignment = align('center', 'center')
    c14.border = thin

# ── ESPACE ────────────────────────────────────────────────────────────────────
ws.row_dimensions[15].height = 10
ws.merge_cells('A15:N15')
ws['A15'].fill = fill(C_SEPARATOR)

# ── LIGNE 16 : SECTION CLASSEMENTS ───────────────────────────────────────────
write_section_header(ws, 'A16', 'A16:N16', "▶  CLASSEMENTS TOP 5 — Meilleures performances par catégorie")
ws.row_dimensions[16].height = 18

def write_top5_block(ws, header_cell, merge_range, title, start_row,
                     start_col_letter, metric_col, col_headers, db_cols, accent_color=C_NAVY):
    write_section_header(ws, header_cell, merge_range, title,
                         style={'fill': fill(accent_color), 'font': font(True, 10, C_WHITE)})

    sc = ord(start_col_letter) - ord('A') + 1

    # Sous-entêtes
    for offset, h in enumerate(col_headers):
        cl = get_column_letter(sc + offset)
        c = ws[f'{cl}{start_row}']
        c.value = h
        apply_style(c, S_SUBHDR)
        c.alignment = align('center', 'center')
        c.border = thin
    ws.row_dimensions[start_row].height = 16

    # Données avec alternance de couleur
    for rank in range(1, 6):
        row = start_row + rank
        ws.row_dimensions[row].height = 15
        match_expr = f"MATCH(LARGE({DR}!{metric_col}:{metric_col},{rank}),{DR}!{metric_col}:{metric_col},0)"
        row_style = S_DATA_ODD if rank % 2 == 1 else S_DATA_EVEN
        for offset, db_col in enumerate(db_cols):
            cl = get_column_letter(sc + offset)
            c = ws[f'{cl}{row}']
            if db_col == metric_col:
                c.value = f"=LARGE({DR}!{metric_col}:{metric_col},{rank})"
            else:
                c.value = f"=INDEX({DR}!{db_col}:{db_col},{match_expr})"
            apply_style(c, row_style)
            c.border = thin
            c.alignment = align('center', 'center', wrap=(offset == 0))

        # Badge rang dans la 1ère colonne
        cl0 = get_column_letter(sc)
        ws[f'{cl0}{row}'].value = rank
        ws[f'{cl0}{row}'].alignment = align('center', 'center')

# Disposition en grille 2×3
write_top5_block(ws, 'A17', 'A17:E17', "🏀  Top 5 Attaquants (PTS)",
                 18, 'A', COL['PTS'],
                 ["#", "Joueur", "Saison", "Équipe", "PTS"],
                 [None, COL['Player'], COL['Season'], COL['Tm'], COL['PTS']],
                 C_NAVY)

write_top5_block(ws, 'G17', 'G17:K17', "🎯  Top 5 Passeurs (AST)",
                 18, 'G', COL['AST'],
                 ["#", "Joueur", "Saison", "Équipe", "AST"],
                 [None, COL['Player'], COL['Season'], COL['Tm'], COL['AST']],
                 C_STEEL)

write_top5_block(ws, 'A25', 'A25:E25', "💪  Top 5 Rebondeurs (TRB)",
                 26, 'A', COL['TRB'],
                 ["#", "Joueur", "Saison", "Équipe", "TRB"],
                 [None, COL['Player'], COL['Season'], COL['Tm'], COL['TRB']],
                 C_TEAL)

write_top5_block(ws, 'G25', 'G25:K25', "🛡️  Top 5 Intercepteurs (STL)",
                 26, 'G', COL['STL'],
                 ["#", "Joueur", "Saison", "Équipe", "STL"],
                 [None, COL['Player'], COL['Season'], COL['Tm'], COL['STL']],
                 C_GREEN)

write_top5_block(ws, 'A33', 'A33:E33', "🚫  Top 5 Contreurs (BLK)",
                 34, 'A', COL['BLK'],
                 ["#", "Joueur", "Saison", "Équipe", "BLK"],
                 [None, COL['Player'], COL['Season'], COL['Tm'], COL['BLK']],
                 C_RED)

# --- Top 5 Défenseurs (STL + BLK) ---
write_section_header(ws, 'G33', 'G33:K33', "🛡️ Top 5 Défenseurs (STL + BLK)",
                     style={'fill': fill(C_RED), 'font': font(True, 10, C_WHITE)})
ws.row_dimensions[33].height = 18

# Sous-entêtes
for h, cl in zip(["#", "Joueur", "Saison", "STL", "BLK"], ['G', 'H', 'I', 'J', 'K']):
    c = ws[f'{cl}34']
    c.value = h
    apply_style(c, S_SUBHDR)
    c.alignment = align('center', 'center')
    c.border = thin
ws.row_dimensions[34].height = 16

# Préparer les données en Python (comme pour les graphiques)
df_def = df.copy()
df_def['def_score'] = df_def['STL'].fillna(0) + df_def['BLK'].fillna(0)  # Somme STL + BLK
top_def = df_def.nlargest(5, 'def_score')[['Player', 'Season', 'STL', 'BLK']].reset_index(drop=True)

# Remplir les données dans Excel
for rank, row in top_def.iterrows():
    r = 34 + rank + 1  # +1 car on commence à la ligne 35
    ws.row_dimensions[r].height = 15
    row_style = S_DATA_ODD if rank % 2 == 0 else S_DATA_EVEN  # rank commence à 0
    ws[f'G{r}'].value = rank + 1
    ws[f'H{r}'].value = row['Player']
    ws[f'I{r}'].value = row['Season']
    ws[f'J{r}'].value = float(row['STL']) if not pd.isna(row['STL']) else 0
    ws[f'K{r}'].value = float(row['BLK']) if not pd.isna(row['BLK']) else 0
    for cl in ['G', 'H', 'I', 'J', 'K']:
        apply_style(ws[f'{cl}{r}'], row_style)
        ws[f'{cl}{r}'].border = thin
        ws[f'{cl}{r}'].alignment = align('center', 'center', wrap=(cl == 'H'))

# --- Top 5 Joueurs avec le plus de MVP (True) ---
write_section_header(ws, 'A41', 'A41:E41', "🏆 Top 5 Joueurs — Plus de MVP (VRAI)",
                     style={'fill': fill(C_GOLD), 'font': font(True, 10, C_MIDNIGHT)})
ws.row_dimensions[41].height = 18

# Sous-entêtes
for h, cl in zip(["#", "Joueur", "Nb MVP", "Saison", "Équipe"], ['A', 'B', 'C', 'D', 'E']):
    c = ws[f'{cl}42']
    c.value = h
    apply_style(c, S_SUBHDR)
    c.alignment = align('center', 'center')
    c.border = thin
ws.row_dimensions[42].height = 16

# Préparer les données en Python
if MVP_c in df.columns:
    # Remplacer les valeurs manquantes par "FAUX" pour éviter les erreurs
    df_mvp = df.copy()
    df_mvp[MVP_c] = df_mvp[MVP_c].fillna("False")

    # Filtrer les joueurs avec MVP = "VRAI"
    mvp_players = df_mvp[df_mvp[MVP_c] == "True"]

    # Compter le nombre de MVP par joueur
    mvp_counts = mvp_players.groupby('Player').size().nlargest(5).reset_index(name='Nb_MVP')

    # Fusionner avec les données pour obtenir la saison et l'équipe
    # Utiliser la première occurrence pour éviter les doublons
    top_mvp = mvp_counts.merge(
        df_mvp[['Player', 'Season', 'Tm']].drop_duplicates('Player'),
        on='Player',
        how='left'
    )

    # Remplir les données dans Excel
    for rank, row in top_mvp.iterrows():
        r = 42 + rank + 1  # +1 car on commence à la ligne 43
        ws.row_dimensions[r].height = 15
        row_style = S_DATA_ODD if rank % 2 == 0 else S_DATA_EVEN
        ws[f'A{r}'].value = rank + 1
        ws[f'B{r}'].value = row['Player']
        ws[f'C{r}'].value = int(row['Nb_MVP'])
        ws[f'D{r}'].value = row['Season'] if not pd.isna(row['Season']) else "N/A"
        ws[f'E{r}'].value = row['Tm'] if not pd.isna(row['Tm']) else "N/A"
        for cl in ['A', 'B', 'C', 'D', 'E']:
            apply_style(ws[f'{cl}{r}'], row_style)
            ws[f'{cl}{r}'].border = thin
            ws[f'{cl}{r}'].alignment = align('center', 'center', wrap=(cl == 'B'))

# --- Top 5 Meilleurs shooteurs 3P (max de 3P) ---
# Utilise write_top5_block avec une plage non fusionnée
write_top5_block(
    ws, 'G41', 'G41:K41', "🎯 Top 5 Shooteurs 3P (3P)",
    42, 'G', COL['3PA'],
    ["#", "Joueur", "Saison", "Équipe", "3P"],
    [None, COL['Player'], COL['Season'], COL['Tm'], COL['3PA']],
    C_STEEL
)

# Ligne finale décorative
last_row = 48
ws.row_dimensions[last_row].height = 6
ws.merge_cells(f'A{last_row}:N{last_row}')
ws[f'A{last_row}'].fill = fill(C_GOLD)

# Ligne finale décorative
last_row = 48
ws.row_dimensions[last_row].height = 6
ws.merge_cells(f'A{last_row}:N{last_row}')
ws[f'A{last_row}'].fill = fill(C_GOLD)

# ── LARGEURS DE COLONNES ─────────────────────────────────────────────────────
col_widths = {
    'A': 5, 'B': 20, 'C': 12, 'D': 10, 'E': 8,
    'F': 3,  # séparateur
    'G': 5, 'H': 20, 'I': 12, 'J': 10, 'K': 8,
    'L': 3,  # séparateur
    'M': 12, 'N': 12,
}
for col_l, w in col_widths.items():
    ws.column_dimensions[col_l].width = w

# Colonnes KPI : uniformes
for i in range(1, 11):
    ws.column_dimensions[get_column_letter(i)].width = 13

# ── AUTOFILTER ────────────────────────────────────────────────────────────────
if "Données brutes" in wb.sheetnames:
    ws_data = wb["Données brutes"]
    ws_data.auto_filter.ref = f"A1:{get_column_letter(ws_data.max_column)}{ws_data.max_row}"

# ═══════════════════════════════════════════════════════════════════════════════
# SAUVEGARDE
# ═══════════════════════════════════════════════════════════════════════════════
wb.save(output_path)
print(f"📁 Fichier sauvegardé : {output_path}")
print("✅ TDB_NBA : Dashboard avec KPIs, Top 5 classiques, et Top 5 spécifiques (Attaquants, Défenseurs, MVP, Shooteurs 3P)")
```
